# Guided analysis

## Step 1: Question

> **How did the daily number of reported street-condition requests differ across New York City's five boroughs in 2025?**


## Step 2: Data

### Import the libraries

Import Pandas, Matplotlib's `pyplot`, and Seaborn using their conventional aliases.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### Read the data

Read `data/nyc_311_street_conditions_2025.csv` into a DataFrame named `requests`. Use `parse_dates=["created_date"]` so Pandas converts the CSV's text timestamps into datetime values while loading the file.


```python
requests = pd.read_csv(
    "data/nyc_311_street_conditions_2025.csv",
    parse_dates=["created_date"]
)

requests
```

### Preview the data

Display the first and last five rows.


```python
display(requests.head())
display(requests.tail())
```

### Inspect the DataFrame

Display its stored data types and missing-value counts.


```python
requests.info()

missing_values = requests.isna().sum()
missing_values
```

### Predict


**Which borough do you predict will have the greatest typical number of reported street-condition requests per day? What informed your prediction?**

## Step 3: Operation


### Make the borough names human-readable

The source stores borough names in all capital letters. **Title case** capitalizes the first letter of each word and writes the remaining letters in lowercase. For example, `STATEN ISLAND` becomes `Staten Island`.

Use `.str.title()` to convert every value in `requests["borough"]` to title case and save the result back into the same feature.


```python
requests["borough"] = requests["borough"].str.title()
```

**What roles do `.str` and `.title()` play in this statement?**

Your answer: .str lets Pandas apply a string operation to every value in the borough column, and .title() changes each borough name to title case.


### Create a calendar date

Create `requests["date"]` with `requests["created_date"].dt.normalize()`. The `.dt` accessor opens Pandas' datetime tools, and `.normalize()` sets the time in every value to midnight while preserving its calendar date.


```python
requests["date"] = requests["created_date"].dt.normalize()
```

### Count requests for each borough and date

Create a DataFrame named `daily_requests` with this sequence:

1. group `requests` by `date` and `borough`;
2. use `.size()` to count rows in every group; and
3. use `.reset_index(name="request_count")` to return a DataFrame whose count column is named `request_count`.

Display the first five rows.


```python
daily_requests = (
    requests
    .groupby(["date", "borough"])
    .size()
    .reset_index(name="request_count")
)

daily_requests.head()
```

**What does one row of `daily_requests` represent? How is that different from one row of `requests`?**

Your answer: One row of daily_requests represents the number of street-condition requests for one borough on one date. This is different from requests, where each row represents one individual 311 service request.


**Why did we remove the time before grouping and counting the requests?** As a hint, try repeating the calculation with `created_date` in place of `date` and inspect the result.

Your answer: We removed the time so all requests from the same day would be grouped together. If we grouped by the full created_date, requests made at different times on the same day would be treated as separate groups.


### Set the borough order

Create `borough_order` containing the five borough names in alphabetical order.


```python
borough_order = [
    "Bronx",
    "Brooklyn",
    "Manhattan",
    "Queens",
    "Staten Island"
]
```

### Calculate the mean daily request count

Create `borough_summary` by grouping `daily_requests` by `borough`, selecting `request_count`, calculating the mean, and using `.reindex(borough_order)`.


```python
borough_summary = (
    daily_requests
    .groupby("borough")["request_count"]
    .mean()
    .reindex(borough_order)
)

borough_summary
```

**Why is comparing only these means incomplete?**

Your answer: Comparing only the means is incomplete because the average does not show how spread out the daily request counts are or the overall shape of each borough's distribution. Two boroughs could have similar averages but still have very different daily patterns.


## Step 4: Check


### Check borough-date coverage

Count the number of rows for each borough in `daily_requests`, then use `.reindex(borough_order)`.


```python
borough_date_counts = (
    daily_requests
    .groupby("borough")
    .size()
    .reindex(borough_order)
)

borough_date_counts
```

**Does every borough have a count for all 365 dates?**

Your answer: Yes. Every borough has 365 rows, so each borough has a request count for every day of the year.


**What would it mean if one or more boroughs had a count of fewer than 365 rows?**

Your answer: It would mean that one or more dates were missing for that borough, so there would not be a daily request count recorded for every day of the year.

### Check the calculated data

Display missing-value counts for `date`, `borough`, and `request_count`. Then use `.describe()` to summarize `request_count` within each borough and put the boroughs in `borough_order`.


```python
daily_requests[["date", "borough", "request_count"]].isna().sum()
daily_requests.groupby("borough")["request_count"].describe().reindex(borough_order)
```

**Are any calculated values missing or impossible?**

Your answer: No. There are no missing calculated values, and the request counts look reasonable. The minimum request count is at least 1 for every borough, so there are no negative or impossible values.


**Would a zero necessarily be impossible in another 311 dataset?**

Your answer: No. A zero could be possible if a borough had no street condition requests on a certain day. It would just mean no requests were reported for that borough on that date.


## Step 5: Evidence


### Create overlapping histograms

Use `sns.histplot()` with:

- `data=daily_requests`;
- `x="request_count"`;
- `hue="borough"` and `hue_order=borough_order`;
- `element="step"`;
- layered histograms;
- `alpha=0.45`; and
- the `"deep"` palette.

Capitalize the legend title and add a question-focused title and descriptive axis labels. Label the y-axis `Number of days`.


```python
ax = sns.histplot(
    data=daily_requests,
    x="request_count",
    hue="borough",
    hue_order=borough_order,
    element="step",
    multiple="layer",
    alpha=0.45,
    palette="deep"
)

plt.title("How Do Daily Street-Condition Requests Differ by Borough?")
plt.xlabel("Daily Number of Requests")
plt.ylabel("Number of days")

ax.get_legend().set_title("Borough")

plt.show()
```

**What does the histogram reveal, and what makes the boroughs easy or difficult to compare?**

Your answer: The histogram shows that Queens generally has the highest daily request counts, followed by Brooklyn, while the Bronx and Staten Island tend to have lower counts. The overlapping colors make it possible to compare the overall shapes, but some of the distributions overlap a lot, which makes exact comparisons harder.

### Create a box plot

Put `request_count` on the x-axis and `borough` on the y-axis. Use `borough_order`, the named color `"mediumseagreen"`, and descriptive labels.


```python
sns.boxplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumseagreen"
)

plt.title("How Do Daily Request Counts Differ by Borough?")
plt.xlabel("Daily Number of Requests")
plt.ylabel("Borough")
plt.show()
```

**What differences among boroughs are easier to see in the box plot than in the histogram?**

Your answer: The box plot makes it easier to compare the median and spread of daily request counts for each borough. It clearly shows that Queens has the highest typical daily count, followed by Brooklyn, and it also makes differences in variability and unusually high values easier to see.

### Create a violin plot

Use the same x feature, y feature, and order. Change only the plot color to `"mediumpurple"` in addition to the required data mappings and order.


```python
sns.violinplot(
    data=daily_requests,
    x="request_count",
    y="borough",
    order=borough_order,
    color="mediumpurple"
)

plt.title("How Do Daily Request Count Distributions Differ by Borough?")
plt.xlabel("Daily Number of Requests")
plt.ylabel("Borough")
plt.show()
```

**What does the violin plot add to the comparison?**

Your answer: The violin plot makes it easier to see the shape of each borough's distribution and where the daily request counts are most concentrated. It also helps show differences in spread and whether some values are more common than others.


**Which of the three figures most clearly compares daily request counts among boroughs? Explain your choice.**

Your answer: I think the box plot compares the boroughs most clearly because it makes the median, spread, and unusually high values easy to compare side by side. The histogram has more overlap, and the violin plot shows shape well, but the box plot is the easiest to read quickly.


## Step 6: Conclusion


**What conclusion about borough differences is supported by the 2025 data?**

Your answer: The 2025 data shows clear differences in daily street condition request counts between boroughs. Queens had the highest typical number of daily requests, followed by Brooklyn, while the Bronx and Staten Island had the lowest. Queens also had the widest spread and some of the highest daily counts.


## Step 7: Limitation


**Why do these request counts not directly measure the number of street problems or borough performance?**

Your answer: These request counts only show how many street condition reports were submitted. They do not show the exact number of street problems because the same problem could be reported more than once, and some problems may never be reported. They also do not directly measure borough performance because the number of requests can be affected by things other than how well the streets are maintained.


# Your own analysis

Develop a new research question using the 311 data. Your question must compare a **quantitative feature across categorical groups**. You will probably need to calculate the quantitative feature from the original request records.

Possible grouping features include `descriptor`, month, or day of the week. You may use another meaningful categorical feature that you can create and explain.


## Step 1: Question


**Write a research question that compares a quantitative feature across categorical groups.**

Your answer: How does the daily number of reported street condition requests differ by day of the week?

## Step 2: Data


**Which original features will you use, and what will each row of your calculated DataFrame represent?**

Your answer: I will use created_date to get the calendar date and day of the week. Each row of my calculated DataFrame will represent one day, including the day of the week and the total number of street-condition requests reported that day.

## Step 3: Operation

Prepare the grouping feature, calculate the quantitative feature, and save the resulting DataFrame with a meaningful name.


```python
requests["day_of_week"] = requests["created_date"].dt.day_name()

weekday_requests = (
    requests
    .groupby(["date", "day_of_week"])
    .size()
    .reset_index(name="request_count")
)

weekday_requests.head()
```

**Briefly explain how your code produced the quantitative feature and comparison groups.**

Your answer: I used created_date to create the day-of-week groups, then grouped the data by date and day of the week. I counted the number of requests in each group to create the daily request_count values.

## Step 4: Check

Check the calculated data for missing or impossible values, confirm that the intended groups are present, and use an appropriate numerical summary to inspect the distributions.


```python
print(weekday_requests.isna().sum())

print(weekday_requests["day_of_week"].value_counts())

weekday_requests.groupby("day_of_week")["request_count"].describe()
```

**What did your checks establish?**

Your answer: The checks showed that all seven days of the week are included and each day has 52 or 53 observations. The request counts are all positive and look reasonable, so I did not find any impossible values in the calculated data.


## Step 5: Evidence

Create a histogram, box plot, or violin plot that directly addresses your question. Include a clear title and descriptive axis labels.


```python
day_order = [
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
    "Sunday"
]

sns.boxplot(
    data=weekday_requests,
    x="request_count",
    y="day_of_week",
    order=day_order
)

plt.title("How Do Daily Street-Condition Requests Differ by Day of the Week?")
plt.xlabel("Daily Number of Requests")
plt.ylabel("Day of the Week")
plt.show()
```

**What pattern in the figure provides the strongest evidence for your answer?**

Your answer: The strongest pattern is that weekday request counts are generally higher than weekend request counts. Tuesday and Wednesday have some of the highest typical daily counts, while Saturday and Sunday are much lower.

## Step 6: Conclusion


**State one conclusion that directly answers your research question.**

Your answer: Daily street condition request counts were generally higher on weekdays than on weekends. Tuesday had the highest average daily request count at about 240.75 requests, while Saturday had the lowest at about 111.52.


## Step 7: Limitation


**State one important limitation on how your result should be interpreted.**

Your answer: The results only show when 311 street condition requests were reported. They do not explain why weekdays had more requests, and they do not prove that there were actually more street problems on those days.
